```python
marks = float(input("Enter your marks (0-100): "))

if not (0 <= marks <= 100):
    print("Please enter marks between 0 and 100.")
else:
    if marks >= 90:
        grade = "A"
    elif marks >= 80:
        grade = "B"
    elif marks >= 70:
        grade = "C"
    elif marks >= 60:
        grade = "D"
    else:
        grade = "F"
    print(f"Your grade is: {grade}")
```
